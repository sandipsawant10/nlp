```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

# Load dataset
df = pd.read_csv("../tilak/Downloads/WineQT.csv")

# Features (remove quality column)
X = df.drop(['quality'], axis=1)

# Scale data (IMPORTANT for clustering)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Model
model = AgglomerativeClustering(n_clusters=3)

# Fit
labels = model.fit_predict(X_scaled)

# Add cluster to dataset
df['cluster'] = labels

print(df[['quality', 'cluster']].head())
```

```python
# average quality per cluster
print(df.groupby('cluster')['quality'].mean())
```
