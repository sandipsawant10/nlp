```python
def viterbi(obs, states, pi, trans, emit):
    V = [{}]
    path = {}

    # init
    for s in states:
        V[0][s] = pi[s] * emit[s][obs[0]]
        path[s] = [s]

    print("After 1 word:", obs[0], "->", max(V[0], key=V[0].get))

    # recursion
    for t in range(1, len(obs)):
        V.append({})
        new_path = {}

        for curr in states:
            (prob, prev) = max(
                (V[t-1][p] * trans[p][curr] * emit[curr][obs[t]], p)
                for p in states
            )

            V[t][curr] = prob
            new_path[curr] = path[prev] + [curr]

        path = new_path

        print("After word", t+1, ":", obs[t], "->", max(V[t], key=V[t].get))

    # final
    final = max(V[-1], key=V[-1].get)
    print("\nFinal Best Sequence:", path[final])
```

```python
states = ['N', 'V']
obs = ['time', 'flies']

pi = {'N': 0.6, 'V': 0.4}

trans = {
    'N': {'N': 0.3, 'V': 0.7},
    'V': {'N': 0.8, 'V': 0.2}
}

emit = {
    'N': {'time': 0.5, 'flies': 0.2},
    'V': {'time': 0.4, 'flies': 0.6}
}

viterbi(obs, states, pi, trans, emit)
```
