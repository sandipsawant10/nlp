```python
def edit_distance(s1, s2):
    m, n = len(s1), len(s2)

    dp = [[0]*(n+1) for _ in range(m+1)]

    # base case
    for i in range(m+1):
        dp[i][0] = i
    for j in range(n+1):
        dp[0][j] = j

    # fill table
    for i in range(1, m+1):
        for j in range(1, n+1):
            if s1[i-1] == s2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(
                    dp[i-1][j],    # delete
                    dp[i][j-1],    # insert
                    dp[i-1][j-1]   # replace
                )

    # backtracking
    i, j = m, n
    ops = []

    while i > 0 and j > 0:
        if s1[i-1] == s2[j-1]:
            i -= 1
            j -= 1
        else:
            if dp[i][j] == dp[i-1][j-1] + 1:
                ops.append(f"Replace {s1[i-1]} → {s2[j-1]}")
                i -= 1
                j -= 1
            elif dp[i][j] == dp[i-1][j] + 1:
                ops.append(f"Delete {s1[i-1]}")
                i -= 1
            else:
                ops.append(f"Insert {s2[j-1]}")
                j -= 1

    # remaining
    while i > 0:
        ops.append(f"Delete {s1[i-1]}")
        i -= 1
    while j > 0:
        ops.append(f"Insert {s2[j-1]}")
        j -= 1

    return dp[m][n], ops[::-1]


# input
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

dist, operations = edit_distance(s1, s2)

print("\nMinimum Edit Distance:", dist)
print("Operations:")
for op in operations:
    print(op)
```
