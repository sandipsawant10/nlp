```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from rake_nltk import Rake

# Load dataset
df = pd.read_csv("../tilak/Downloads/IMDB Dataset.csv")

# Convert sentiment to 0/1
df['sentiment'] = df['sentiment'].map({'positive':1, 'negative':0})

X = df['review']
y = df['sentiment']

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Vectorize
cv = CountVectorizer(stop_words='english')
X_train = cv.fit_transform(X_train)
X_test = cv.transform(X_test)

# Train model
model = MultinomialNB()
model.fit(X_train, y_train)

# -------- TEST CASE --------
r = Rake()

sample = X_test.iloc[0]

# Sentiment prediction
sample_vec = cv.transform([sample])
pred = model.predict(sample_vec)[0]

# Convert label to text
sentiment = "Positive" if pred == 1 else "Negative"

# Keywords
r.extract_keywords_from_text(sample)
keywords = r.get_ranked_phrases()[:5]

# Output
print("Review:\n", sample[:150], "...\n")
print("Sentiment:", sentiment)
print("Keywords:", keywords)
```
