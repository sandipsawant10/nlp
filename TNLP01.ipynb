```python
import pandas as pd
import re 
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
df = pd.read_csv("../tilak/Downloads/sentimentdataset.csv")
```

```python
df.head()
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text=df['Text']
```

```python
stop_words=set(stopwords.words('english'))
```

```python
def preprocess(text):
    text=text.lower()
    text=re.sub(r'[^a-z\s]' , '' , text)
    tokens=word_tokenize(text)
    tokens=[w for w in tokens if w not in stop_words]
    return tokens 
```

```python
df['preprocessed_text']=text.apply(preprocess)
```
