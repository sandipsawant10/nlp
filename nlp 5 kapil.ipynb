```python
from fastcoref import FCoref
import re
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
model = FCoref(device='cpu')
```

```python
text = """
Rohit went to the market. He bought apples.
Later, Rohit met Suresh and he invited him for dinner.
"""
```

```python
preds = model.predict(texts=[text])
result = preds[0]
```

```python
print(preds[0])
```

```python
clusters = result.get_clusters(as_strings=True)
```

```python
print("Coreference Clusters:")
for cluster in clusters:
    print(cluster)
```

```python
resolved_text = text
```

```python
for cluster in clusters:
    main_entity = cluster[0]   # First mention = main entity
    
    for mention in cluster[1:]:
        # Replace ONLY whole words (avoid breaking 'the')
        pattern = r'\b' + re.escape(mention) + r'\b'
        resolved_text = re.sub(pattern, main_entity, resolved_text)

print("\nResolved Text:")
print(resolved_text)
```
