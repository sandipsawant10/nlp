```python
import nltk
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

s1 = "I went to the bank to deposit money"
s2 = "The river overflowed the bank"

# Sentence 1
words1 = word_tokenize(s1)
sense1 = lesk(words1, "bank")

# Sentence 2
words2 = word_tokenize(s2)
sense2 = lesk(words2, "bank")

print("Sentence 1:", s1)
print("Sense:", sense1)
print("Meaning:", sense1.definition())

print("\nSentence 2:", s2)
print("Sense:", sense2)
print("Meaning:", sense2.definition())
```

```python
from nltk.corpus import wordnet as wn

def wsd(sentence, word):
    words = set(word_tokenize(sentence.lower()))
    best = None
    max_overlap = 0

    for sense in wn.synsets(word):
        meaning = set(word_tokenize(sense.definition().lower()))
        overlap = len(words & meaning)

        if overlap > max_overlap:
            max_overlap = overlap
            best = sense

    return best


s1 = "I went to the bank to deposit money"
s2 = "The river overflowed the bank"

sense1 = wsd(s1, "bank")
sense2 = wsd(s2, "bank")

print("Sentence 1:", s1)
print("Sense:", sense1)
print("Meaning:", sense1.definition())

print("\nSentence 2:", s2)
print("Sense:", sense2)
print("Meaning:", sense2.definition())
```
